# Hyperinstanced MEP Models to Improve Memory Consumption of Large BIM Scenes

MEP layers of a BIM model tend to be very large. We previously were able to compress down the size of this file by replacing all the unique pieces of geometry with one mesh that was scaled, rotated and translated such that it looked like the sample object. We applied this hyperinstancing method on the straight pipes of the scene and were able to compress the file size down by a factor of 94%. However, we did not include the fittings in this file, and essentially stripped it down to just vertices and faces.

While straight pipes could be compressed down very easily, this was essentially low-hanging fruit. We want to reach the very top of the tree here, and that will involve instancing our fittings too - things like valves, elbows, tees etc. Identifying straight pipe also relied heavily on proper metadata associations with the object itself, and that will not always be the case.

Our goal through this workbook will be to create a reproducible pipeline that is able to classify and hyperinstance our scene.
