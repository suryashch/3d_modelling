# IFC Data Retrieval

The goal of this notebook is to explore concepts of retrieving data from an IFC file using python and `ifcopenshell`.

```python
import pandas as pd
import ifcopenshell
import ifcopenshell.util.element
import ifcopenshell.util.classification
```

```python
model = ifcopenshell.open('data/O-S1-INS-Plumbing - Sanitair.ifc')
```

The first thing I'd like to explore is retrieving actors by GUID. Here, we have a list of test GUIDs which might get retrieved from the model.

```python
test_uids = [
    "1TAKhTtRj1m9w5JPRxJsWi",
    "2I4NgicpDCAxSK_gG5HVpI",
    "2QnWhfdtj9rxv0_kyM8jOI",
    "1TAKhTtRj1m9w5JPRxJsWw",
    "0e27GKX4fEUOi_YGLe2dh0",
    "2I4NgicpDCAxSK_gG5HVpQ"
]
```

From the [docs](https://docs.ifcopenshell.org/ifcopenshell-python/geometry_processing.html), we see that `model.by_guid()` is a method which may work.

```python
model.by_guid(test_uids[0])
```

We see that information is indeed retrieved. Let's see if we can parse this into something useable. First, lets save this result to a variable.

```python
test_element = model.by_guid(test_uids[0])
```

Now, lets try getting the property sets, type and other attributes of this element.

```python
test_element_type = ifcopenshell.util.element.get_type(test_element)
test_element_type
```

```python
ifcopenshell.util.element.get_psets(test_element)
```

```python
ifcopenshell.util.element.get_quantities(test_element_type)
```

```python
ifcopenshell.util.element.get_properties(test_element_type)
```

```python
test_element.get_info()
```

```python
test_element_type.get_info()
```

```python
test_element.is_a()
```

```python
ifcopenshell.util.classification.get_references(test_element)
```

Looks good for now. Let's see if we can do some aggregations.

```python
for uid in test_uids:
    element = model.by_guid(uid)

    print(element.is_a())
```

```python
for uid in test_uids:
    element = model.by_guid(uid)

    print(element.get_info())
```

```python
psets_headers = set()

for uid in test_uids:
    element = model.by_guid(uid)

    psets = ifcopenshell.util.element.get_psets(element)

    for pset in psets.keys():
        for header in psets[pset]:
            psets_headers.add(header)

print(psets_headers)
```
